> What aspects of design did you consider when making your visualization? How did you apply them? With what elements of your plots? 

    I started with cleaning up the dataset to remove duplicate agengy and null data. I also created a new clean data with only columns of data that could be useful for my visualization.
    I tried to clarify the graph by simplifying the x-axis labels by removing 
    the "Eligibility" prefix from each category (e.g., "EligibilityAdult" to "Adult"). Removing visual clutter.
    Instead of estimating the values from the y-axis, I added the count labels to allow exact values to be read at a glance. This reduces additional effort required to interpret the graph.

    In th excel version, I have also done the cleaning process using pivot tables on the clean dataset to quickly get the data required (eligibility criteria number) to create as similar visual as with python.
    Each bar represents a single eligibility category (Adult, Child, 
    Family, Female, Male, Teen). In python, a unique color was assigned to each bar to visually distinguish each category at a glance
    Bar width and spacing were kept consistent to allow fair visual comparison across all categories.
    


```python
import plotly.express as px

df_map = df_clean.dropna(subset=['Latitude', 'Longitude'])

fig = px.scatter_map(
    df_map,
    lat='Latitude',
    lon='Longitude',
    hover_name='ResourceAgencyNum',
    hover_data=['PublicName', 'MailingPostalCode'],
    color_discrete_sequence=['purple'],
    zoom=10,
    height=600
)

fig.update_layout(
    map_style='open-street-map',
    title=dict(
        text='<b>Mental Health Agency Locations</b>',
        x=0.5,
        xanchor='center'
    ),
    margin=dict(r=0, t=40, l=0, b=0)
)

fig.show()
```

```python
df_map = df_clean.dropna(subset=['Latitude', 'Longitude'])

# Create readable services summary column
def get_services(row):
    services = []
    if row['EligibilityAdult'] == True: services.append('Adult')
    if row['EligibilityChild'] == True: services.append('Child')
    if row['EligibilityFamily'] == True: services.append('Family')
    if row['EligibilityFemale'] == True: services.append('Female')
    if row['EligibilityMale'] == True: services.append('Male')
    if row['EligibilityTeen'] == True: services.append('Teen')
    return ', '.join(services) if services else 'Not Specified'

df_map = df_map.copy()
df_map['Eligibility'] = df_map.apply(get_services, axis=1)

# Filter to Ontario boundaries only
df_map = df_map[
    (df_map['Latitude'] >= 41.6) & (df_map['Latitude'] <= 56.9) &
    (df_map['Longitude'] >= -95.2) & (df_map['Longitude'] <= -74.3)
]

# Format hover template for clean readable box
df_map['hover_text'] = (
    '<b>Agency:</b> ' + df_map['PublicName'].fillna('N/A') + '<br>' +
    '<b>Postal Code:</b> ' + df_map['MailingPostalCode'].fillna('N/A') + '<br>' +
    '<b>Eligibility:</b> ' + df_map['Eligibility']
)

fig = px.scatter_map(
    df_map,
    lat='Latitude',
    lon='Longitude',
    hover_name=None,
    hover_data={
        'Latitude': False,
        'Longitude': False,
        'Eligibility': False,
        'MailingPostalCode': False,
        'hover_text': False
    },
    color='Eligibility',
    color_discrete_sequence=px.colors.qualitative.Safe,
    zoom=5,
    center={"lat": 51.2538, "lon": -85.3232},
    height=700,
)

# Apply custom hover template
fig.update_traces(
    hovertemplate=(
        '<b style="font-size:13px">%{customdata[0]}</b>'
    ),
    customdata=df_map[['hover_text']].values,
    hoverlabel=dict(
        bgcolor='white',
        bordercolor='#0072B2',
        font=dict(
            size=12,
            color='black',
            family='Arial'
        ),
        namelength=0
    ),
    marker=dict(
        size=8,
        opacity=0.8
    )
)

fig.update_layout(
    map_style='open-street-map',
    title=dict(
        text='<b>Mental Health Agency Locations in Ontario (2019)</b>',
        x=0.5,
        xanchor='center',
        font=dict(size=18, color='black')
    ),
    showlegend=False,
    margin=dict(r=0, t=60, l=0, b=40),
    map=dict(
        bounds=dict(
            west=-95.2,
            east=-74.3,
            south=41.6,
            north=56.9
        )
    )
)

fig.add_annotation(
    text="💡 Hover over each point to view agency name, postal code, and eligibility services",
    xref="paper", yref="paper",
    x=0.5, y=-0.04,
    showarrow=False,
    font=dict(size=11, color='grey'),
    xanchor='center'
)

fig.show()
```

### 2 Geographic Distribution

 > What information or message are you trying to convey with your visualization? 
The map show the physical location of each mental health agency. This visualization helps identify:
- Where agencies in densely populated urban areas
- Gaps in service coverage in rural or underserved regions
- The mental health agencies across different neighborhoods

> What aspects of design did you consider when making your visualization? How did you apply them? With what elements of your plots? 
    
I tried different version to improve clarity and make sure the audience can get essential information. 
- I created a second map that is centered on Ontario since the data is only covering that province and it creates more focus on the relevant region without distraction from surrounding areas. It also prevents the audience from panning outside Ontario, keeping the focus on relevant geographic region.
- I also wanted the audience to have a display the entire province at once, giving 
  an immediate overview of agency distribution.
- Hover boxes were designed to display only relevant information: Agency Name, Postal Code, and Eligibility Services. The goal was to keep the hover information concise and meaningful showing only what a user needs to know at a glance.
I decided to not add a legend to reduce visual clutter on the map since eligibility information is accessible through the hover interaction.
After with the support of chat GPT, I was looking to make the point accessible and it was recommended to set the point size to 8 with 80% opacity to ensure agencies are visible without overlapping points obscuring each other in densely populated urban areas like Toronto.
